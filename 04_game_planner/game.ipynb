## State

This is the cell you keep editing. Add a `research(...)` as you finish one, a `have(...)` when you
put down miners, and a `module.*` line for everything you build.

```python
f = Factory(name='my-game')

f.research('steam-power', 'electronics', 'automation-science-pack', 'automation', 'logistics')

f.have('iron-ore', 120)      # per minute, from miners
f.have('copper-ore', 60)

f.add(
    module.iron_plate(120),
    module.copper_plate(60),
    module.gear(30),
    module.red_science(10),
)
f
```

## The next move

`f.next()` reads the ladder, the research and the rates and gives you one thing to do.
`f.apply(move)` takes it: a build becomes a module in the factory, a research is marked done,
a supply is added — the same as typing the line yourself in the state cell.

```python
move = f.next()
move
```

```python
f.apply(move)
f.status()
```

## Throughput and upgrades

A module is fixed hardware: the machines its rate needed, in as many columns as the belts of the
day required. `throughput()` is what that hardware can do now — the lesser of what the machines
craft and what the belts carry — so a tier upgrade changes it without rebuilding anything.

```python
g = module.gear(900)
g
```

```python
g.throughput(), g.capacity(), g.port_limit(), g.limited_by()
```

```python
g.upgrade('assembling-machine-3')       # faster machines, same ground
g.throughput(), g.limited_by()          # now the belts are the limit
```

```python
g.upgrade('fast-transport-belt')        # belts catch up
g.throughput(), g.limited_by()
```

```python
# what it draws at the rate it is asked to run, and what is left in the tank
g.inputs(), g.output(), g.headroom()
```

```python
# shrink it back to the rate you asked for, now that the tiers are better
g.rebuild()
g
```
